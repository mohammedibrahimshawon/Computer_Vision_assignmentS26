### Conversion between pixel and camera coordinates
Load data set `ueb24.npy` `(np.load)`.
The image $B(i,j)$ is not a conventional intensity image. Each pixel $p = (x,y,f)^T$ encodes the distance from the focal point of the camera $(0, 0, 0)^T$ to a point in the 3D scene $P = (X, Y, Z)$. $p$ is the projection of $P$ on the image plane. So, the location corresponding to the pixel $p$ from the image matrix $B(i,j)$ will hold the distance $r$ between the point $P$ and the focal point. The image plane containing the pixel $p$, is placed at the distance $f$ from the focal point, parallel to the $XY$ plane.

Write a Python routine that maps each pixel $p$ in $B(i,j)$ to the corresponding point $P = (X,Y,Z)$. Visualize the result in a 3D surface plot `(plt.plot_surface)`. Consider that you are using a camera with focal length $f = 0.008$ meters and a CCD-chip, in which the size of a pixel is $0.00004$ meters. The optical axis intersects the centre of the image matrix, which corresponds to the frame coordinates $(72.5, 88.5)$ on the CCD-chip.

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
from matplotlib import cm

# ── Load data ──────────────────────────────────────────────────────────────
B = np.load('data/exercise_2/ueb24.npy')          # shape (145, 177); each pixel stores distance r

# ── Camera parameters ──────────────────────────────────────────────────────
f        = 0.008      # focal length [m]
pixel_s  = 0.00004   # pixel size on CCD [m]

# ── Image dimensions ───────────────────────────────────────────────────────
rows, cols = B.shape          # (145, 177)
cx, cy = 88.5, 72.5          # optical axis centre in frame coordinates (col, row)
# NOTE: B(i,j) → i is row index (Y direction), j is col index (X direction)

# ── Pixel grid in camera coordinates ───────────────────────────────────────
# j runs over columns → x_cam; i runs over rows → y_cam
j_idx = np.arange(cols)      # 0 … 176
i_idx = np.arange(rows)      # 0 … 144

J, I = np.meshgrid(j_idx, i_idx)   # each (145, 177)

# Camera-plane coordinates [m]  (centre subtracted, scaled by pixel size)
x_cam = (J - cx) * pixel_s   # horizontal offset from optical axis
y_cam = (I - cy) * pixel_s   # vertical offset from optical axis

# ── Direction vector from focal point to each pixel p ──────────────────────
# p = (x_cam, y_cam, f)^T  — the image plane is at distance f along Z
# Unit vector: p_hat = p / |p|
p_norm = np.sqrt(x_cam**2 + y_cam**2 + f**2)   # |p|

# Distance r stored in B
r = B   # r = distance from focal point (0,0,0) to scene point P

# ── Back-project to 3-D ────────────────────────────────────────────────────
# P = r * (p / |p|) = r * p_hat
X = r * (x_cam / p_norm)
Y = r * (y_cam / p_norm)
Z = r * (f     / p_norm)
```

```python
# if you're not using the notebook in Colab, you can use "%matplotlib notebook"
# to create an interactive plot that can be rotated
%matplotlib inline
def plot_3d(x, y, z, elev, azim):
  fig = plt.figure()
  ax = fig.add_subplot(111, projection='3d')
  ax.view_init(elev=elev, azim=azim)
  ax.plot_surface(x,y,-z,cmap=cm.coolwarm)


plot_3d(X, Y, Z, 45, -45)
plot_3d(X, Y, Z, 90, -90)
plot_3d(X, Y, Z, 0, 0)
```
